```python
df_clean = df.copy()

print("Working copy created.")
print("Original shape:", df.shape)
print("Working copy shape:", df_clean.shape)
```

```python
# Replace whitespace-only strings with NaN
df_clean["TotalCharges"] = (
    df_clean["TotalCharges"]
    .astype(str)
    .str.strip()
    .replace("", np.nan)
)

# Convert to numeric
df_clean["TotalCharges"] = pd.to_numeric(
    df_clean["TotalCharges"],
    errors="coerce"
)

print(df_clean["TotalCharges"].dtype)
```

```python
print("Missing TotalCharges:",
      df_clean["TotalCharges"].isnull().sum())

print("Total missing values:",
      df_clean.isnull().sum().sum())
```

```python
output_path = "../data/processed/telco_churn_cleaned.csv"

df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
```

```python
total_customers = len(df_clean)

churned_customers = (df_clean["Churn"] == "Yes").sum()
retained_customers = (df_clean["Churn"] == "No").sum()

churn_rate = (churned_customers / total_customers) * 100
retention_rate = (retained_customers / total_customers) * 100

print(f"Total Customers: {total_customers}")
print(f"Churned Customers: {churned_customers}")
print(f"Retained Customers: {retained_customers}")
print(f"Churn Rate: {churn_rate:.2f}%")
print(f"Retention Rate: {retention_rate:.2f}%")
```

```python
# Check categorical values
categorical_columns = df_clean.select_dtypes(
    include="object"
).columns

for col in categorical_columns:
    print(f"\n{col}:")
    print(df_clean[col].value_counts(dropna=False))
```

```python
# Invalid numerical values
print("Negative tenure:",
      (df_clean["tenure"] < 0).sum())

print("Negative MonthlyCharges:",
      (df_clean["MonthlyCharges"] < 0).sum())

print("Negative TotalCharges:",
      (df_clean["TotalCharges"] < 0).sum())

print("Zero tenure with non-missing TotalCharges:",
      ((df_clean["tenure"] == 0) &
       (df_clean["TotalCharges"].notna())).sum())
```
