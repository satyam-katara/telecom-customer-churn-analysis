# Telecom Customer Churn — Retention Strategy

## Objective

Develop practical, data-driven customer retention strategies
based on the findings from SQL analysis, Python EDA,
and statistical testing.

## Focus Areas

1. New Customer Retention
2. Fiber Optic Customer Retention
3. Electronic Check Customer Retention

## Expected Outcome

Recommend actionable retention initiatives with measurable
KPIs and pilot testing plans.

## 1. New Customer Retention Strategy

### Business Problem
New customers are more likely to leave the company, particularly those using month-to-month contracts.

### Data Evidence
- New customers: 2,186
- Churned new customers: 1,037
- New customer churn rate: 47.44%
- New month-to-month customers: 1,994
- Churned new month-to-month customers: 1,024
- Segment churn rate: 51.35%
- Share of total churn from this segment: 54.79%

### Recommended Action
Introduce an early-stage customer retention program for new month-to-month customers.

The program should include:
- Proactive customer check-ins during the first 90 days.
- Early identification of billing and service-related complaints.
- Clear communication about available support services.
- An optional incentive for customers considering a longer-term contract.

### Success KPIs
- 90-day customer retention rate.
- Monthly churn rate among new customers.
- Customer complaint resolution time.
- Conversion rate from month-to-month to longer-term contracts.

### Pilot Plan
Run a controlled pilot with eligible new month-to-month customers. Compare the churn rate of customers receiving the retention program against a comparable control group.

Evaluate results after 90 days before expanding the program.

### Expected Business Impact
The strategy aims to reduce early customer churn and protect recurring monthly revenue. Its effectiveness should be measured through the pilot rather than assumed in advance.

## 2. Fiber Optic Customer Retention Strategy

### Business Problem
Fiber optic customers show a considerably higher churn rate than customers using DSL or no internet service.

### Data Evidence
- Fiber optic customers: 3,096
- Churned fiber customers: 1,297
- Fiber optic churn rate: 41.89%
- Monthly charges associated with churned fiber customers: $114,300.05
- Share of total churned customers: 69.40%

### Recommended Action
Introduce a targeted retention and service-quality review program for fiber optic customers.

The program should include:
- Monitor customer complaints and service issues.
- Review internet speed and connection reliability.
- Improve technical support response time.
- Contact dissatisfied customers before contract cancellation.
- Offer suitable retention plans based on customer needs.

### Success KPIs
- Fiber optic customer churn rate.
- Average complaint resolution time.
- Number of unresolved service complaints.
- Customer satisfaction score.
- Monthly recurring revenue retained.

### Pilot Plan
Select eligible fiber optic customers and divide them into treatment and control groups.

Provide proactive service monitoring and improved support to the treatment group. Compare churn rates between both groups over a defined period.

### Expected Business Impact
The initiative aims to improve fiber customer experience, reduce churn and protect recurring revenue.

The dataset identifies higher churn among fiber customers, but does not establish the underlying service-quality cause. Further investigation and pilot testing are required.

## 3. Electronic Check Customer Retention Strategy

### Business Problem
Customers using electronic check payments have a higher churn rate compared to customers using automatic bank transfer, credit card or mailed check.

### Data Evidence
- Electronic check customers: 2,365
- Churned customers: 1,071
- Churn rate: 45.29%
- Monthly charges associated with churned customers: $84,288.75
- Share of total churned customers: 57.30%

### Recommended Action
Investigate the payment experience of electronic check customers and identify possible billing-related issues.

The initiative should include:
- Review payment-related complaints and failed transactions.
- Identify billing confusion and payment difficulties.
- Improve payment reminders and billing communication.
- Make automatic payment options easier to understand and use.
- Offer alternative payment methods without forcing customers to switch.

### Success KPIs
- Electronic check customer churn rate.
- Payment failure rate.
- Billing-related complaint rate.
- Successful payment rate.
- Customer adoption of automatic payment methods.

### Pilot Plan
Run a controlled pilot among eligible electronic check customers.

Provide improved billing communication and easier access to alternative payment options to the treatment group. Compare churn and payment-related outcomes against a control group over a defined period.

### Expected Business Impact
The initiative aims to improve the payment experience, reduce billing-related friction and retain recurring revenue.

The dataset shows an association between electronic check usage and higher churn, but does not establish that the payment method itself causes customers to leave. Further investigation and pilot testing are required.
